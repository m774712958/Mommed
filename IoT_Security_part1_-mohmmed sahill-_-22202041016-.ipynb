```python
# Filename: IoT_Security_Simulation_Part1.py
import json
import random
import base64
from cryptography.fernet import Fernet

# --- Part I: IoT Device Data Encryption Simulation ---
# 1. Setup Encryption Key (Simulating a pre-shared key)
key = Fernet.generate_key()
cipher_suite = Fernet(key)

def generate_sensor_data():
    """Generates random temperature and humidity readings."""
    return {
        "device_id": "SENSOR_001",
        "temperature": round(random.uniform(20.0, 35.0), 2),
        "humidity": round(random.uniform(40.0, 60.0), 2)
    }

def simulate_iot_transmission():
    print("-" * 40)
    print("PART I: IoT Data Encryption Simulation")
    print("-" * 40)

    # Step 1: Generate Data
    data = generate_sensor_data()
    json_data = json.dumps(data)
    print(f"Original Data (Cleartext): {json_data}")

    # Step 2: Encrypt Data (using AES/Fernet)
    # Encoding to bytes is required for encryption
    encrypted_data = cipher_suite.encrypt(json_data.encode('utf-8'))
    print(f"Encrypted Data (Transmitted): {encrypted_data}")

    # Step 3: Simulate Transmission & Decrypt
    print("\n[Network Transmission...]\n")

    # Receiver Side Decryption
    decrypted_bytes = cipher_suite.decrypt(encrypted_data)
    decrypted_data = decrypted_bytes.decode('utf-8')

    print(f"Decrypted Data (Received): {decrypted_data}")
    print("" * 40)

if __name__ == "__main__":
    simulate_iot_transmission()
```
