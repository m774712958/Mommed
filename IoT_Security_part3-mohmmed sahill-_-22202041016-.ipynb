```python

# Filename: IoT_Security_Simulation_Part3.py
import hashlib

# --- Part III: Secure Device Boot Verification (Simulation) ---

def calculate_hash(firmware_content):
    """
    Calculates the SHA-256 hash of the firmware content.
    This simulates checking the digital signature of the code.
    """
    return hashlib.sha256(firmware_content.encode('utf-8')).hexdigest()

def secure_boot(stored_signature, current_firmware):
    print("\n" + "=" * 40)
    print("PART III: Secure Boot Verification")
    print("=" * 40)

    print(f"Stored Trusted Signature: {stored_signature}")

    # [span_0](start_span)Step 2: Verify the hash matches the stored signature[span_0](end_span)
    current_hash = calculate_hash(current_firmware)
    print(f"Current Firmware Hash:    {current_hash}")

    if current_hash == stored_signature:
        # If match, allow boot
        print("\n[SUCCESS] Integrity Verified. System Booting...")
        print("Running Main Loop...")
    else:
        # [span_1](start_span)Step 3: If fails, refuse to boot[span_1](end_span)
        print("\n[ALARM] SECURITY ALERT! Firmware has been tampered with.")
        print("SYSTEM HALTED. Boot Refused.")
    print("=" * 40)

# --- Simulation Execution ---

if __name__ == "__main__":
    # 1. [span_2](start_span)Setup: Define the original valid firmware and store its hash[span_2](end_span)
    original_firmware_code = "System_Version_1.0_Official_Release"
    factory_trusted_signature = calculate_hash(original_firmware_code)

    # Scenario A: Normal Boot (No Attack)
    print("\n--- Scenario A: Normal Boot ---")
    secure_boot(factory_trusted_signature, original_firmware_code)

    # Scenario B: Attack Simulation (Modified Firmware)
    print("\n--- Scenario B: Malware/Tampered Firmware ---")
    infected_firmware = "System_Version_1.0_Official_Release_WITH_MALWARE"
    secure_boot(factory_trusted_signature, infected_firmware)
```
