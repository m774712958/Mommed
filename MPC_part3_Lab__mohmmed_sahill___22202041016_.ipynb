```python
## Part III: Conceptual Garbled Circuits (AND Gate)

# Inputs
A = 1 # Omar's Input
B = 0 # Nancy's Input
Expected_Output = A and B # 1 AND 0 = 0

# 1. Setup: Key/Label Generation (Conceptual)
# Instead of AES keys, we use simple descriptive labels.
Labels = {
    'A0': 'K_A_Zero', 'A1': 'K_A_One',
    'B0': 'K_B_Zero', 'B1': 'K_B_One',
    'C0': 'K_C_Zero', 'C1': 'K_C_One' # Output Keys
}

# 2. Garbling the AND Gate Table (Encryption is simplified)
# The output key K_C_C is conceptually encrypted with input keys K_A and K_B.
def garble_cell(input_key_A, input_key_B, output_key):
    # Conceptual Encryption: Combines the output key with the input keys used for decryption
    return f"ENC({input_key_A}, {input_key_B})_OUT_{output_key}"

Garbled_Circuit_Table = {
    # (A, B) -> C
    '00': garble_cell(Labels['A0'], Labels['B0'], Labels['C0']), # 0 AND 0 = 0
    '01': garble_cell(Labels['A0'], Labels['B1'], Labels['C0']), # 0 AND 1 = 0
    '10': garble_cell(Labels['A1'], Labels['B0'], Labels['C0']), # 1 AND 0 = 0
    '11': garble_cell(Labels['A1'], Labels['B1'], Labels['C1'])  # 1 AND 1 = 1
}

print("\n--- Part III: Garbled Circuit (Conceptual AND Gate) ---")
print("Garbled Table (Conceptual):")
for k, v in Garbled_Circuit_Table.items():
    print(f"  {k}: {v}")

# 3. Evaluation Phase
# Omar gets his label (A=1)
Omar_Label = Labels[f'A{A}']
# Nancy gets her label (B=0) via Oblivious Transfer (simulated)
Nancy_Label = Labels[f'B{B}']

print(f"\nOmar's Label: {Omar_Label} | Nancy's Label: {Nancy_Label}")

# The Evaluator (Nancy) tries to decrypt all 4 entries with the pair (Omar_Label, Nancy_Label).
found_output_key = None

for entry in Garbled_Circuit_Table.values():
    # Only ONE entry will "decrypt" successfully.
    # We simulate successful decryption by checking if the input keys are present in the conceptual encryption.
    if Omar_Label in entry and Nancy_Label in entry:
        # Extract the conceptual output key from the entry string
        found_output_key = entry.split('_OUT_')[1]
        break

# 4. Decryption and Output
if found_output_key == Labels['C0']:
    Final_Output = 0
elif found_output_key == Labels['C1']:
    Final_Output = 1
else:
    Final_Output = "Error in evaluation"

print(f"\nDecrypted Output Key: {found_output_key}")
print(f"Final Computed Output (A AND B): {Final_Output}")
print(f"Expected Output: {Expected_Output}")
```
