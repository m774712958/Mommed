```python
# Filename: IoT_Security_Simulation_Part2.py
import time
import datetime

# --- Part II: IoT Device Lifecycle Simulation ---

def log_event(stage, message):
    """Helper to print timestamped logs."""
    timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    print(f"[{timestamp}] [Stage {stage}] {message}")
    time.sleep(1.5) # Simulate processing time

def simulate_lifecycle():
    print("\n" + "-" * 40)
    print("PART II: IoT Device Security Lifecycle")
    print("-" * 40)

    # Stage 1: Threat Modeling
    log_event(1, "Threat model created. Vulnerabilities identified.")

    # Stage 2: Secure Boot Initialization
    log_event(2, "Secure boot verified. Firmware signature matches.")

    # Stage 3: Secure Key Injection
    # Simulating injecting a mock key
    mock_key = "A1B2-C3D4-E5F6-SECURE"
    log_event(3, f"Keys injected securely. Root Key ID: {hash(mock_key)}")

    # Stage 4: OTA Firmware Update Check
    log_event(4, "Checking for OTA updates... No new firmware found. System Integrous.")

    # Stage 5: Secure Decommissioning
    log_event(5, "Decommissioning initiated. Wiping storage...")
    log_event(5, "Secrets wiped. Device safely decommissioned.")
    print("-" * 40)

if __name__ == "__main__":
    simulate_lifecycle()
```
