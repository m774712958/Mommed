```python
## Part II: Simplified Shamir's Secret Sharing (2, 3)

SECRET = 100
K = 2 # Threshold: Need 2 shares to reconstruct
N = 3 # Total shares generated

# 1. Share Generation (The Dealer)
# Use a simple linear polynomial: P(x) = a0 + a1*x
a0 = SECRET # The secret is the y-intercept (P(0))
import random
a1 = random.randint(1, 10) # Random coefficient (slope)

print("\n--- Part II: Shamir's Secret Sharing (Simplified) ---")
print(f"Polynomial: P(x) = {a0} + {a1}*x")

shares = []
for x in range(1, N + 1): # x coordinates are 1, 2, 3
    y = a0 + a1 * x
    shares.append((x, y))
    print(f"Share {x}: (x={x}, y={y})")

# 2. Secret Reconstruction (Using K=2 shares)
# We use Shares 1 and 3: (x1, y1) and (x2, y2)
# The slope a1 = (y2 - y1) / (x2 - x1)
# The secret a0 = y1 - a1 * x1

x1, y1 = shares[0] # Share 1
x2, y2 = shares[2] # Share 3

print(f"\nUsing shares 1 ({x1}, {y1}) and 3 ({x2}, {y2}) for reconstruction.")

# Calculate the slope (a1)
# NOTE: In a real system, division requires Modular Inverse
slope_a1 = (y2 - y1) / (x2 - x1)

# Calculate the secret (a0)
secret_reconstructed = y1 - slope_a1 * x1

print(f"Reconstructed Slope (a1): {slope_a1}")
print(f"Reconstructed Secret:     {int(secret_reconstructed)}") # Convert back to int
print(f"Original Secret:          {SECRET}")

# Q5 Test (Failure with fewer than K=2 shares)
print("\nQ5 Test: Trying to reconstruct with only 1 share...")
x_fail, y_fail = shares[0] # Only Share 1
# With only one point, the line (polynomial) is NOT uniquely determined.
print(f"Cannot reconstruct the secret with only one share ({x_fail}, {y_fail}).")
```
