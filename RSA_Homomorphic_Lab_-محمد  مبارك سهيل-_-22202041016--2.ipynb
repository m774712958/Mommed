```python

def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

def mod_inverse(e, phi):
    for d in range(3, phi):
        if (d * e) % phi == 1:
            return d
    return None

def encrypt(m, e, n):
    return pow(m, e, n)

def decrypt(c, d, n):
    return pow(c, d, n)

# Step 1: Key Generation
p, q = 11, 13
n = p * q
phi = (p - 1) * (q - 1)

e = 7
d = mod_inverse(e, phi)

print(f"p = {p}, q = {q}")
print(f"n = {n}, φ = {phi}")
print(f"Public key (e, n) = ({e}, {n})")
print(f"Private key (d, n) = ({d}, {n})")

# Step 2: Encryption
m1, m2 = 5, 9
c1 = encrypt(m1, e, n)
c2 = encrypt(m2, e, n)

print(f"m1 = {m1}, m2 = {m2}")
print(f"E(m1) = {c1}, E(m2) = {c2}")

# Step 3: Homomorphic Multiplication
c_product = (c1 * c2) % n
decrypted = decrypt(c_product, d, n)
expected = (m1 * m2) % n

print(f"E(m1)*E(m2) mod n = {c_product}")
print(f"Decrypted result = {decrypted}")
print(f"Expected (m1*m2 mod n) = {expected}")

if decrypted == expected:
    print("Homomorphic property verified: E(m1)E(m2) ≡ E(m1*m2) mod n")
else:
    print("Homomorphic property failed")
```
