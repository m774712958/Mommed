```python
## Part I: Simplified MPC Sum using Additive Secret Sharing

# 1. Setup: Private Inputs
Ahmed_Private_Input = 50
Ali_Private_Input = 70
Rashida_Private_Input = 100
Expected_Total = Ahmed_Private_Input + Ali_Private_Input + Rashida_Private_Input # 220

# 2. Sharing Phase: Each party generates random shares.
# We will use simple integers instead of modular arithmetic.

def simple_share(secret):
    # Generates 3 shares (s1, s2, s3) such that s1 + s2 + s3 = secret
    import random
    r1 = random.randint(1, 100)
    r2 = random.randint(1, 100)
    r3 = secret - r1 - r2  # The last share closes the sum
    return [r1, r2, r3]

Ahmed_shares = simple_share(Ahmed_Private_Input) # [A1, A2, A3]
Ali_shares = simple_share(Ali_Private_Input)     # [B1, B2, B3]
Rashida_shares = simple_share(Rashida_Private_Input) # [C1, C2, C3]

print("--- Part I: MPC Sum (Additive Sharing) ---")
print(f"Ahmed's Shares: {Ahmed_shares}")
print(f"Ali's Shares:   {Ali_shares}")
print(f"Rashida's Shares: {Rashida_shares}")
# Note: No party knows the other's private input (50, 70, 100).

# 3. Local Computation: Each party sums the shares they received.
# Ahmed computes S1 = A1 + B1 + C1
partial_sum_P1 = Ahmed_shares[0] + Ali_shares[0] + Rashida_shares[0]
# Ali computes S2 = A2 + B2 + C2
partial_sum_P2 = Ahmed_shares[1] + Ali_shares[1] + Rashida_shares[1]
# Rashida computes S3 = A3 + B3 + C3
partial_sum_P3 = Ahmed_shares[2] + Ali_shares[2] + Rashida_shares[2]

print(f"\nPartial Sums: P1={partial_sum_P1}, P2={partial_sum_P2}, P3={partial_sum_P3}")

# 4. Reconstruction: All parties reveal their partial sums.
Global_Sum = partial_sum_P1 + partial_sum_P2 + partial_sum_P3

print(f"\nComputed Global Sum: {Global_Sum}")
print(f"Expected Total:      {Expected_Total}")
```
